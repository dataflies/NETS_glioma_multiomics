```python
import os
import csv

# ===============================
# 1. 定义路径
# ===============================
mri_root = r"E:\NETS\影像数据\0_CGGA.MRI_268"                     # 核磁数据根目录
gene_file = r"E:\NETS\bulk rna数据/CGGA.mRNAseq_325.RSEM-genes.20200506.txt"  # 基因表达文件
clinical_file = r"E:\NETS\bulk rna数据\CGGA.mRNAseq_325_clinical.20200506.txt"  # 原始临床数据
output_clinical = r"E:\NETS\bulk rna数据\CGGA.mRNAseq_325_clinical_filtered.txt"  # 输出文件
output_gene = r"E:\NETS\bulk rna数据\CGGA.mRNAseq_filtered_genes.txt"          # 输出：筛选后的基因表达矩阵
# ===============================
# 2. 获取基因数据中的所有患者ID
# ===============================
print("读取基因数据...")
with open(gene_file, 'r', encoding='utf-8') as f:
    reader = csv.reader(f, delimiter='\t')
    header = next(reader)          # 第一行：Gene_Name + 患者ID列表
    all_gene_patients = header[1:]  # 所有在基因数据中有表达的患者ID
print(f"基因数据中共有 {len(all_gene_patients)} 个患者")

# ===============================
# 3. 扫描MRI文件夹，筛选同时满足两个条件的患者
# ===============================
valid_patients = []
print("扫描MRI文件夹...")
for item in os.listdir(mri_root):
    item_path = os.path.join(mri_root, item)
    if os.path.isdir(item_path):
        patient_id = item
        if patient_id not in all_gene_patients:
            continue
        t2_path = os.path.join(item_path, "T2_bet.nii.gz")
        ce_path = os.path.join(item_path, "CE_bet.nii.gz")
        if os.path.isfile(t2_path) and os.path.isfile(ce_path):
            valid_patients.append(patient_id)
            print(f"✓ 纳入: {patient_id}")
        else:
            print(f"✗ 跳过: {patient_id} (缺少T2或CE)")

print(f"\n最终纳入患者数: {len(valid_patients)}")

# ===============================
# 4. 提取临床数据中对应患者的行
# ===============================
print("正在提取临床数据...")
with open(clinical_file, 'r', encoding='utf-8') as fin, \
     open(output_clinical, 'w', encoding='utf-8', newline='') as fout:
    reader = csv.reader(fin, delimiter='\t')
    writer = csv.writer(fout, delimiter='\t')
    
    # 写入表头
    header_clinical = next(reader)
    writer.writerow(header_clinical)
    
    # 逐行处理
    matched_count = 0
    for row in reader:
        if row and row[0] in valid_patients:
            writer.writerow(row)
            matched_count += 1
    
print(f"已保存 {matched_count} 条临床记录到: {output_clinical}")
print("完成！")

print("正在提取基因表达数据...")
# 先确定需要保留的患者列在原表头中的索引
with open(gene_file, 'r', encoding='utf-8') as f:
    reader = csv.reader(f, delimiter='\t')
    original_header = next(reader)
    # 需要保留的列索引：[0] 是基因名，再加上 valid_patients 对应的列
    keep_cols = [0] + [i for i, pid in enumerate(original_header[1:], start=1) if pid in valid_patients]

# 逐行读取并写入新文件
with open(gene_file, 'r', encoding='utf-8') as fin, \
     open(output_gene, 'w', encoding='utf-8', newline='') as fout:
    reader = csv.reader(fin, delimiter='\t')
    writer = csv.writer(fout, delimiter='\t')
    for row in reader:
        new_row = [row[i] for i in keep_cols]
        writer.writerow(new_row)

print(f"基因表达矩阵已保存至: {output_gene}")
```

```python
import os
import csv
import shutil

# ===============================
# 1. 定义路径（请根据你的实际路径修改）
# ===============================
mri_root = r"E:\NETS\影像数据\0_CGGA.MRI_268"                     # MRI 根目录
gene_file = r"E:\NETS\bulk rna数据/CGGA.mRNAseq_325.RSEM-genes.20200506.txt"
clinical_file = r"E:\NETS\bulk rna数据\CGGA.mRNAseq_325_clinical.20200506.txt"

output_gene = r"E:\NETS\bulk rna数据\CGGA.mRNAseq_filtered_genes.txt"
output_clinical = r"E:\NETS\bulk rna数据\CGGA.mRNAseq_693_clinical_filtered.txt"
output_mri = r"E:\NETS\影像数据\filtered_MRI"   # 影像输出文件夹

# ===============================
# 2. 获取基因数据中的所有患者ID
# ===============================
print("读取基因数据表头...")
with open(gene_file, 'r', encoding='utf-8') as f:
    reader = csv.reader(f, delimiter='\t')
    header = next(reader)
    all_gene_patients = header[1:]   # 第一列是 Gene_Name
print(f"基因数据中共有 {len(all_gene_patients)} 个患者")

# ===============================
# 3. 扫描MRI文件夹，筛选同时满足两个条件的患者
# ===============================
valid_patients = []
print("扫描MRI文件夹...")
for item in os.listdir(mri_root):
    item_path = os.path.join(mri_root, item)
    if os.path.isdir(item_path):
        patient_id = item
        if patient_id not in all_gene_patients:
            continue
        t2_path = os.path.join(item_path, "T2_bet.nii.gz")
        ce_path = os.path.join(item_path, "CE_bet.nii.gz")
        if os.path.isfile(t2_path) and os.path.isfile(ce_path):
            valid_patients.append(patient_id)
            print(f"✓ 纳入: {patient_id}")
        else:
            print(f"✗ 跳过: {patient_id} (缺少T2或CE)")

print(f"\n最终纳入患者数: {len(valid_patients)}")

if len(valid_patients) == 0:
    raise SystemExit("没有符合条件的患者，程序终止。")

# ===============================
# 4. 提取基因表达矩阵（只保留有效患者列）
# ===============================
print("正在提取基因表达数据...")
# 先确定需要保留的列索引
with open(gene_file, 'r', encoding='utf-8') as f:
    reader = csv.reader(f, delimiter='\t')
    original_header = next(reader)
    # 需要保留的列索引：[0]是基因名，再加上valid_patients对应的列
    keep_cols = [0] + [i for i, pid in enumerate(original_header[1:], start=1) if pid in valid_patients]

# 逐行写入新文件
with open(gene_file, 'r', encoding='utf-8') as fin, \
     open(output_gene, 'w', encoding='utf-8', newline='') as fout:
    reader = csv.reader(fin, delimiter='\t')
    writer = csv.writer(fout, delimiter='\t')
    for row in reader:
        new_row = [row[i] for i in keep_cols]
        writer.writerow(new_row)

print(f"基因表达矩阵已保存到: {output_gene}")

# ===============================
# 5. 提取临床数据（只保留有效患者行）
# ===============================
print("正在提取临床数据...")
with open(clinical_file, 'r', encoding='utf-8') as fin, \
     open(output_clinical, 'w', encoding='utf-8', newline='') as fout:
    reader = csv.reader(fin, delimiter='\t')
    writer = csv.writer(fout, delimiter='\t')
    header_clinical = next(reader)
    writer.writerow(header_clinical)
    matched_count = 0
    for row in reader:
        if row and row[0] in valid_patients:
            writer.writerow(row)
            matched_count += 1

print(f"临床数据已保存到: {output_clinical} (共 {matched_count} 条)")

# ===============================
# 6. 复制影像文件到新文件夹
# ===============================
print("开始复制影像文件...")
os.makedirs(output_mri, exist_ok=True)
copied_count = 0

for patient_id in valid_patients:
    src_dir = os.path.join(mri_root, patient_id)
    dst_dir = os.path.join(output_mri, patient_id)
    os.makedirs(dst_dir, exist_ok=True)
    
    t2_src = os.path.join(src_dir, "T2_bet.nii.gz")
    ce_src = os.path.join(src_dir, "CE_bet.nii.gz")
    t2_dst = os.path.join(dst_dir, "T2_bet.nii.gz")
    ce_dst = os.path.join(dst_dir, "CE_bet.nii.gz")
    
    if os.path.isfile(t2_src):
        shutil.copy2(t2_src, t2_dst)
        print(f"✓ 复制 T2: {patient_id}")
    else:
        print(f"✗ 警告：{patient_id} 缺少 T2_bet.nii.gz")
        
    if os.path.isfile(ce_src):
        shutil.copy2(ce_src, ce_dst)
        print(f"✓ 复制 CE: {patient_id}")
    else:
        print(f"✗ 警告：{patient_id} 缺少 CE_bet.nii.gz")
    
    copied_count += 1

print(f"\n影像文件复制完成！共处理 {copied_count} 个患者，影像保存在: {output_mri}")
print("全部任务完成！")
```
